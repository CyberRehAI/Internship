```python
from pathlib import Path
import sys, zipfile, subprocess

WINDOWS_DIR = Path("/content/drive/MyDrive/AI-TDP/windows")
CODE_ZIP = Path("/content/drive/MyDrive/AI-TDP/baselines_code.zip")
WINDOWS_ZIP = Path("/content/drive/MyDrive/AI-TDP/baselines_windows.zip")
REPO_ROOT = Path("/content/AI-TDP")
OUT_ROOT = Path("/content/drive/MyDrive/AI-TDP/baselines/outputs")
MODE = "net"  # or "concat"

subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "torch", "numpy", "scikit-learn", "matplotlib"])
REPO_ROOT.mkdir(parents=True, exist_ok=True)
if not (REPO_ROOT / "baselines" / "models").is_dir():
    with zipfile.ZipFile(CODE_ZIP, "r") as zf:
        zf.extractall(REPO_ROOT)
if not (WINDOWS_DIR / "train.npz").is_file():
    WINDOWS_DIR.mkdir(parents=True, exist_ok=True)
    with zipfile.ZipFile(WINDOWS_ZIP, "r") as zf:
        zf.extractall(WINDOWS_DIR)
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))
print("Ready", REPO_ROOT, WINDOWS_DIR)
```

```python
import runpy
OUT_DIR = OUT_ROOT / f"lstm_ae_{MODE}"
OUT_DIR.mkdir(parents=True, exist_ok=True)
sys.argv = [
    "train_lstm_ae.py",
    "--mode", MODE,
    "--windows-dir", str(WINDOWS_DIR),
    "--out-dir", str(OUT_DIR),
    "--epochs", "50",
    "--patience", "8",
]
runpy.run_path(str(REPO_ROOT / "baselines" / "train" / "train_lstm_ae.py"), run_name="__main__")
```
